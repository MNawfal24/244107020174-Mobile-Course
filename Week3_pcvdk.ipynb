```python
import cv2 as cv
import matplotlib.pyplot as plt

path_ktm1a = '/content/drive/MyDrive/Pcvdk_Nawfal/KTM_Gelap.jpeg'
path_ktm1b = '/content/drive/MyDrive/Pcvdk_Nawfal/KTM_luar_ruangan.jpeg'

img_ktm1a = cv.imread(path_ktm1a)
img_ktm1b = cv.imread(path_ktm1b)

# Konversi BGR ke RGB
img_ktm1a_rgb = cv.cvtColor(img_ktm1a, cv.COLOR_BGR2RGB)
img_ktm1b_rgb = cv.cvtColor(img_ktm1b, cv.COLOR_BGR2RGB)

# Konversi Invers
inv_ktm1a = 255 - img_ktm1a_rgb
inv_ktm1b = 255 - img_ktm1b_rgb

# Hasil Inver Ruangan Gelap
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(img_ktm1a_rgb)
plt.title("KTM Asli \n(Ruangan Gelap)")

plt.subplot(1, 2, 2)
plt.imshow(inv_ktm1a)
plt.title(f"{NIM} - Inverse KTM Gelap \n(Menjadi Pucat/Berkabut)")
plt.show()

# Hasil Invers Luar Ruangan
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(img_ktm1b_rgb)
plt.title("KTM Asli \n(Luar Ruangan)")

plt.subplot(1, 2, 2)
plt.imshow(inv_ktm1b)
plt.title(f"{NIM} - Inverse KTM1b \n(Tetap Kontras)")
plt.show()
```
